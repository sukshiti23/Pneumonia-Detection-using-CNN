# **Pneumonia Detection using Convolutional Neural Network (CNN)**

## Project Overview
This project uses a Convolutional Neural Network (CNN) to classify chest X-ray images as either *Normal* or *Pneumonia*. The model is trained on grayscale chest X-ray images and learns image features such as edges, textures, and patterns to make predictions.

## Dataset
- Chest X-ray images
- Two classes:
  - Normal
  - Pneumonia
- Images are resized to *150 × 150* pixels.
- Images are converted to grayscale before training.

## Model Architecture
The CNN consists of:
- Rescaling layer for image normalization
- Convolutional (Conv2D) layers with ReLU activation
- MaxPooling layers
- Flatten layer
- Dense hidden layer
- Output layer with Sigmoid activation for binary classification


## Training
The model is trained using:
- Optimizer: Adam
- Loss Function: Binary Crossentropy
- Evaluation Metric: Accuracy

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
# Data Prepaeation
DATA_DIR = "chest_xray"
IMAGE_SIZE = (150, 150)
BATCH_SIZE = 32
```

```python
# Load the dataset
train_data = tf.keras.preprocessing.image_dataset_from_directory(
    DATA_DIR + "/train",
    image_size=IMAGE_SIZE,
    batch_size = BATCH_SIZE,
    color_mode = "grayscale"
)
```

```python
val_data = tf.keras.preprocessing.image_dataset_from_directory(
    DATA_DIR + "/val",
    image_size=IMAGE_SIZE,
    batch_size = BATCH_SIZE,
    color_mode = "grayscale"
)
```

```python
test_data = tf.keras.preprocessing.image_dataset_from_directory(
    DATA_DIR + "/test",
    image_size=IMAGE_SIZE,
    batch_size = BATCH_SIZE,
    color_mode = "grayscale"
)
```

```python
print("Classes = ", train_data.class_names)
```

```python
# BUILDING THE CNN

model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(150, 150, 1)), # (size of the image, number of channels)
    # Normalizing pixel values from 0-255 to 0-1
    tf.keras.layers.Rescaling(1./255),

    # Convolutional layer 
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'), # 32 filters, 3x3 kernel size
    # Pooling layer
    tf.keras.layers.MaxPooling2D((2,2)), # reduces the spatial dimensions of the output from the convolutional layer
    # Convolutional layer 
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'), # 64 filters, 3x3 kernel size
    # Pooling layer
    tf.keras.layers.MaxPooling2D((2,2)), # reduces the spatial dimensions of the output from the convolutional layer
    # Convolutional layer 
    tf.keras.layers.Conv2D(128, (3, 3), activation='relu'), # 128 filters, 3x3 kernel size
    # Pooling layer
    tf.keras.layers.MaxPooling2D((2,2)), # reduces the spatial dimensions of the output from the convolutional layer

    # Flattening to 1D
    tf.keras.layers.Flatten(),
    # Fully connected layer
    tf.keras.layers.Dense(128, activation='relu'),

    # Optimising
    tf.keras.layers.Dropout(0.5), # Dropout layer to prevent overfitting    

    # Output layer
    tf.keras.layers.Dense(1, activation='sigmoid')
])

```

```python
# Compiling the model
model.compile(
    optimizer='adam',
    loss = "binary_crossentropy",
    metrics = [
        "accuracy",
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall')
        ]
)
```

```python
model.summary() # Display the model architecture
```

```python
# Training the model
history = model.fit(
    train_data,
    validation_data = val_data,
    epochs = 3,
    verbose = 2
    )
```

```python
probabilities = model.predict(test_data)
y_pred = (probabilities >= 0.5).astype(int)
```

```python
results = model.evaluate(test_data)

print("Loss:", results[0])
print("Accuracy:", results[1])
print("Precision:", results[2])
print("Recall:", results[3])
```

```python
if y_pred[0][0] < 0.5:
    print("Normal")
else:
    print("Pneumonia Detected")
```

```python
# saving the model
model.save("my_model.keras")
```
