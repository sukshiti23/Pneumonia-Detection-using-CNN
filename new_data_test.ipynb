```python
import tensorflow as tf
import numpy as np
```

```python
# Load the saved model
model = tf.keras.models.load_model("my_model.keras")
```

```python
# loading a new image
IMG_PATH = "new_data/new_image.jpeg"
img = tf.keras.preprocessing.image.load_img(
    IMG_PATH,
    color_mode = "grayscale",
    target_size=(150,150)
)
```

```python
# Convert the image to a numpy array
img_array = tf.keras.preprocessing.image.img_to_array(img)

img_array = img_array/255.0 # normalizing between 0 and 1

img_array = np.expand_dims(img_array, axis=0) # used to add a batch dimension because TensorFlow models expect input in batches, even if you're predicting only one image. 
```

```python
# Predicting
pred = model.predict(img_array)

if pred[0][0] > 0.5:
    print("Pneumonia Detected")
else:
    print("Normal")
```
